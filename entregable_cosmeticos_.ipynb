```python
# pandas

# 1. importar pandas

import pandas as pd
```

```python
# 2. importar JSON

url = "https://www.datos.gov.co/resource/sji7-3uxf.json"

df_json = pd.read_json(url)
print(df_json)
```

```python
# 3. ver las primeras filas

print(df_json.head()) # muestra las primeras 5 lineas
```

```python
# 4. ver las ultimas filas

print(df_json.tail()) # - - - - muestra las ultimas 5 lineas
```

```python
# 5. ver nombres de columnas

print(df_json.columns)
```

```python
# 6. ver indices

print(df_json.index)
```

```python
# 7. seleccionar una columna

print(df_json["razon_social"])
```

```python
# 8. seleccionar varias columnas

print(df_json[["razon_social", "nit", "email_comercial"]])
```

```python
# 9. seleccionar una fila

print(df_json.loc[1])
```

```python
# 10. seleccionar filas con columnas

print(df_json.loc[[1, 2, 3], ["razon_social", "nit", "email_comercial"]])
```

```python
# 11. filtrar datos

mun_comercial = df_json[df_json["mun_comercial"] == "PEREIRA"]
print(mun_comercial)
```

```python
# 12. Agregar una nueva columna

df_json["pais"] = "Colombia"
print(df_json)
```

```python
# 13. eliminar columnas

df_json = df_json.drop("organizacion", axis=1)
print(df_json)
```

```python
# 14. eliminar filas

df_json = df_json.drop([1, 2], axis=0)
print(df_json)
```

```python
# 15. ordenar datos

order = df_json.sort_values("mun_comercial")
print(order)
```

```python
# 16. ordenar antiguedad

order = df_json.sort_values("fec_matricula", ascending=False)
print(order)
```

```python
# 17. cambiar el nombre de las columnas

df_json = df_json.rename(columns={"mun_comercial": "ciudad", "email_comercial": "E_MAIL"})
print(df_json)
```

```python
# 18. contar registros

count = df_json["ciudad"].value_counts()
print(count)
```

```python
# 19. agrupar datos

group = df_json.groupby("ciudad")["razon_social"].count()
print(group)
```

```python
# 20. contar registros

count = df_json["razon_social"].value_counts()
print(count)
```

```python
# 21. guardar en csv

df_json.to_csv("data_json.csv")
```
