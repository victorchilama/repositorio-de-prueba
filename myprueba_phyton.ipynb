```python
# pandas

# 1. importar pandas

import pandas as pd
```

```python
# 2. crear una serie

serie = pd.Series([10,20,30,40,50])
print(serie)
```

```python
# 3. crear dataframe

data = {
    'Nombre': ['Juan', 'Ana', 'Pedro', 'Maria'],
    'Edad': [25, 30, 35, 40],
    'Ciudad': ['Madrid', 'Barcelona', 'Valencia', 'Sevilla']
}

df = pd.DataFrame(data)
print(df)
```

```python
# 4. ver las primeras filas

print(df.head()) # muestra las primeras 5 filas
```

```python
# 6 informacion del dataframe

print(df.info())
```

```python
# 7 estadistica basica

print(df.describe())
```

```python
# 8. ver nombres de columnas

print(df.columns)
```

```python
# 15. agregar una nueva columna

df['Salario'] = ['2000', '3000', '4000', '5000']
print(df)
```

```python
# 19. ordenar datos

order = df.sort_values(by='Edad', ascending=False)
print(order)
```

```python
# 24. guardar en csv

df.to_csv('data.csv', index=False)
```

```python
# 25. leer un archivo en csv

new_df = pd.read_csv('data.csv')
print(new_df)
```
